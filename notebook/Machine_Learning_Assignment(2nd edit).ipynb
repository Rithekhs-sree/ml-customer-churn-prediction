## Dataset Loading

The dataset used in this project is the Telco Customer Churn
dataset provided through IBM's public data repository.

The dataset contains customer-related demographic, service,
contract, and billing information that can be used to investigate
customer churn.

Source: IBM
Dataset: Telco_customer_churn.xlsx

```python
import pandas as pd

df = pd.read_excel("Telco_customer_churn.xlsx")

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)

df.head()
```

```python
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
df.info()
```

```python
missing_summary = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Missing Percentage": (
        df.isnull().sum().values / len(df) * 100
    )
})

missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)
```
